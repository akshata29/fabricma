# nb_06 — DimTerritory

**Purpose:** Load `DimTerritory` from `Files/seed_data/dim_territory.csv` into a Delta table.

## Overview
- Reads `dim_territory.csv`
- Writes to Delta table `DimTerritory` (overwrite)
- No SCD2 — territory data is stable reference data

**Prerequisite:** nb_00_setup_lakehouse must be run first.

```python
# Read seed CSV
df = spark.read.csv(
    "Files/seed_data/dim_territory.csv",
    header=True,
    inferSchema=True
)

print(f"Rows read: {df.count()}")
df.printSchema()
df.show(10)
```

```python
# Write to Delta table
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("DimTerritory")
)

print("DimTerritory loaded successfully.")
spark.sql("SELECT COUNT(*) AS row_count FROM DimTerritory").show()
```
