# nb_02 — DimCurrency

**Purpose:** Load the `DimCurrency` dimension from `Files/seed_data/dim_currency.csv` into a Delta table.

## Overview
- Reads `dim_currency.csv` with header inference
- Writes to Delta table `DimCurrency` (overwrite)
- No SCD2 needed — currency data is stable reference data

**Prerequisite:** nb_00_setup_lakehouse must be run first.

```python
# Read the seed CSV
df = spark.read.csv(
    "Files/seed_data/dim_currency.csv",
    header=True,
    inferSchema=True
)

print(f"Rows read: {df.count()}")
df.printSchema()
df.show(5)
```

```python
# Write to Delta table
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("DimCurrency")
)

print("DimCurrency loaded successfully.")
spark.sql("SELECT COUNT(*) AS row_count FROM DimCurrency").show()
```
