# nb_00 — Set Up MeridianSupplyCo Lakehouse

**Purpose:** Configure the Fabric Lakehouse for the MeridianSupplyCo analytical workload.

## Overview
- Enable Delta Lake extensions on the Spark session
- Set the default database/schema to `meridian`
- Verify that seed CSV files are available under `Files/seed_data/`
- Print session configuration summary

**Run this notebook first** before executing any dimension or fact notebooks.

```python
# Configure Delta Lake extensions
spark.conf.set("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
spark.conf.set("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
spark.conf.set("spark.databricks.delta.schema.autoMerge.enabled", "true")
```

```python
# Create and use the meridian schema (database)
spark.sql("CREATE DATABASE IF NOT EXISTS meridian")
spark.sql("USE meridian")
print("Default database set to: meridian")
```

```python
# Verify seed CSV files exist under Files/seed_data/
import subprocess
seed_files = [
    "Files/seed_data/dim_currency.csv",
    "Files/seed_data/dim_customer.csv",
    "Files/seed_data/dim_product.csv",
    "Files/seed_data/dim_salesrep.csv",
    "Files/seed_data/dim_supplier.csv",
    "Files/seed_data/dim_territory.csv",
    "Files/seed_data/dim_warehouse.csv",
    "Files/seed_data/fact_invoice_header.csv",
    "Files/seed_data/fact_invoice_lines.csv",
    "Files/seed_data/fact_sales.csv",
    "Files/seed_data/fact_stock_movement.csv",
]

for f in seed_files:
    try:
        df_check = spark.read.csv(f, header=True)
        print(f"OK  {f}  ({df_check.count()} rows)")
    except Exception as e:
        print(f"MISSING  {f}  — {e}")
```

```python
# Print Spark session info
print("Spark version:", spark.version)
print("App name:", spark.sparkContext.appName)
print("Default database:", spark.catalog.currentDatabase())
print("Setup complete — proceed to dimension notebooks.")
```
