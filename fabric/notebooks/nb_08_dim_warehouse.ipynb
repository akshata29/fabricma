# nb_08 — DimWarehouse

**Purpose:** Load `DimWarehouse` from `Files/seed_data/dim_warehouse.csv` into a Delta table.

## Overview
- Reads `dim_warehouse.csv`
- Writes to Delta table `DimWarehouse` (overwrite)
- No SCD2 — warehouse data is stable reference data

**Prerequisite:** nb_00_setup_lakehouse must be run first.

```python
# Read seed CSV
df = spark.read.csv(
    "Files/seed_data/dim_warehouse.csv",
    header=True,
    inferSchema=True
)

print(f"Rows read: {df.count()}")
df.printSchema()
df.show(10)
```

```python
# Write to Delta table
(
    df.write
    .format("delta")
    .mode("overwrite")
    .option("overwriteSchema", "true")
    .saveAsTable("DimWarehouse")
)

print("DimWarehouse loaded successfully.")
spark.sql("SELECT COUNT(*) AS row_count FROM DimWarehouse").show()
```
